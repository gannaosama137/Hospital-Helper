```python
from pathlib import Path
import pandas as pd 
```

```python
ROOT = Path.cwd().parent 
DATA_DIR = ROOT / "Data" 
file_path = DATA_DIR / "cardio_train.csv"
df = pd.read_csv(file_path , sep=';') 
```

```python
def explore_data(df):  

    print (f"Shape : {df.shape}" ) 

    print ("\nFirst 5 rows :" ) 
    display(df.head(5)) 

    print ("\ninfo :" )
    df.info()

    print ("\ndescribe :" ) 
    display(df.describe())

    print ("Missin values :\n" , df.isnull().sum() ) 

    print("Duplicated rows :\n" , df.duplicated().sum() )

    print ("\ncolumns :\n" , df.columns )

    return df
```

```python
df = explore_data(df) 
```

```python
categorical_cols = [
    'gender', 'cholesterol', 'gluc',
    'smoke', 'alco', 'active'
]
numerical_cols = [
    'age', 'height', 'weight', 'ap_hi', 'ap_lo'
]

for col in categorical_cols:
    # print(f"\n{col}:")
    print(df[col].value_counts(normalize=True) * 100)
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

 
    outliers = df[(df[col] <lower) | (df[col]>upper)]
    print(f"{col}: Number of Outliers = {outliers.shape[0]}")
    print(f"Min outlier:\n {outliers[col].min()}")
    print(f"Max outlier:\n {outliers[col].max()}")
    print('\n')
```
